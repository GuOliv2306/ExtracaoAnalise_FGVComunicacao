```python
import requests 
import json
import os
import time
import pandas as pd
```

```python
url_base = 'https://uselessfacts.jsph.pl/api/v2/facts/random'
list_fatos = []

for _ in range(10):  # Collect 10 unique facts
    try:
        response = requests.get(url_base)
        response.raise_for_status()
        print('Conexão bem sucedida')
        dados = response.json()
        # Check if this fact's id is already in list_fatos
        if not any(fato['id'] == dados['id'] for fato in list_fatos):
            list_fatos.append(dados)
    except Exception as exc:
        print('Ocorreu um problema: %s' % (exc))
```

```python
dados = list_fatos
df = pd.DataFrame(dados)
df
```
